```python
#prod
import numpy as np
#1D
a=np.array([1,2,3,4])
b=np.array([1,5,8,9])
print(np.prod(a,axis=0))
#2D
a1=np.array([[1,2,3,4],[5,2,1,3]])
a2=np.array([[1,3,3,4],[5,2,1,3]])
print(np.prod(a1,axis=0)) #columns wise
print(np.prod(a1,axis=1))  #rows wise
print(np.prod((a1,a2),axis=1))
print(np.prod((a1,a2),axis=0))
#3D
a3=np.array([[[1,2,3,4]],
             [[1,5,6,2]]])
a4=np.array([[[1,2,3,4]],
             [[1,5,6,2]]])

print(a3.ndim)
print(np.prod(a3,axis=1))
print(np.prod((a3,a4),axis=0))
print(np.prod((a3,a4),axis=1))
```

```python
#static operation
a=np.array([1,8,5,2,1,3,2,1])
print(np.mean(a))
print(np.median(a)) #it create the number from again ascendning order then check median
print(np.var(a))
print(np.std(a))
```

```python
#dot product operation
#it is not valid for only one array
#it valid for 2 arrays
#it is valid when columns of the first matrix==rows of the second matrix
a=np.array([[2,3],[2,3],[1,5]])  #3 rows 2 columns
b=np.array([[1,2,3,4],[1,3,4,1]]) # 2 rows 4 columns
print(np.dot(a,b)) #reslt will be 3 rows and 4 columns
```

```python
#exponential and log function
a=np.array([1,2,3,4])
print(np.exp(a))
print(np.log(a))
```

```python
#indexing and slicing

#1d array
#indexing
a1=np.array([1,2,3,4])
print(a[-4])
#slicing
print(a1[0:3])
print(a1[2:]) #it print all the elements after index of 2 and that included(2)
print(a1[:3])
print(a1[:])  #nothing will change stay as same like copy it

#2d array
a2=np.array([[2,3],[2,4],[1,5]])
#indexing and slicing
print(a2[0])    #0th rows
print(a2[0,1])  #0th rows in first columns
print(a2[0 ,:]) #perticular rows accessing
print(a2[:,1])  #perticular columns accessing
print(a2[0:2:2])
print(a2[::-1])  #reversing
print(a2[1:3])
print(a2[1:3,1])

#3d array
a3=np.array([[[1,2,3,4]],
             [[1,5,6,2]]])
#slicing and indexing
print(a3[-1])
print(a3[0,0,2])
print(a3[0:1])
print(a3[0:2,0:2,1:])
```

```python
#transpose
a=np.array([1,2,3,4])
print(np.transpose(a))  #in 1d array treanspose is same 
a2=np.array([[2,3],[2,4],[1,5]])
print(np.transpose(a2))
a3=np.array([[[1,2,3,4]],
             [[1,5,6,2]]])
print(np.transpose(a3))
```

```python
#ravel it convert 2d and 3d array into 1d array
a3=np.array([[[1,2,3,4]],
             [[1,5,6,2]]])
print(np.ravel(a3))
```

```python
#stack : combine two same array it give another  new dimension or another array
a=np.array([1,2,3,4])
b=np.array([1,2,3,5])
print(np.stack((a,b)))

a2=np.array([[2,3],[2,4],[1,5]])
a3=np.array([[2,3],[2,4],[1,5]])
print(np.stack((a2,a3)))

print(np.hstack((a2,a3))) #horizontal stack
print(np.vstack((a2,a3)))  #vertical stack
```
