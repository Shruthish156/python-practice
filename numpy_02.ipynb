```python
#random number
import numpy as np
a=np.random.random((2,2)) # it automatically crete in float type
print(a)
#for int
b=np.random.randint(1,11,size=5)  #size it create how many elements we want to print
print(b)
```

```python
#linear space : it give equal spaced values b/w two numbers (start,stop,number)
import numpy as np
a=np.linspace(0,100,6)
b=np.linspace(20,1,5) #it give 5 numbers 
print(a,b)
```

```python
#array operations 
import numpy as np
#1.scaler operation : means working with one vector and scaler that is scaler operation
a=np.array([1,2,3,4])
print(a+2)
print(a-2)
print(a*2)
print(a/2)
print(a**2)  #square
print(a**3)  #cube
#relational
print(a==2)
#in the above example "a" is vector and "2" is scaler

#2. vector operation :means working with two vectors
b=np.array([2,5,6])
c=np.array([5,4,2])
#arithmetic
print(b+c)
print(b-c)
print(b*c)
print(b/c)

#relational
print(b>c)
```

```python
#array function
a=np.array([[1,2,6,5],[5,2,5,3]])
print(a.sum())
print(a.min())
print(a.max())
print(a.prod())
print(a.mean()) #average
print(np.prod(a,axis=0))  #columns
print(np.prod(a,axis=1))  #rows
print(np.sin(a))
print(np.cos(a))
```
