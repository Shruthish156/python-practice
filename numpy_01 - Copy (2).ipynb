```python
#1d array matrix is nothing but we called as vector
import numpy as np 
a=np.array([4,5,8,9])
b=np.array([5,8,9,7])
print(a+b) 
```

```python
# 2d array matrix
import numpy as np
a=np.array([[1,5,6,7],[4,8,9,7]])
print(a)
print(a.ndim)   #it use to check the dimension
print(a.shape)  #it give how many rows and colums present
print(a.size)   #it check th etotal elements or numbers present in array
```

```python
#3d array
import numpy as np
a=np.array([[[1,8,9,2],[8,9,5,7],[5,8,9,1]],[[5,8,9,4],[8,9,1,4],[8,9,1,4]]])
print(a)
print(a.ndim)
print(a.shape)
```

```python
#convert given array matrix into float or int
a=np.array([[[1,8,9,2],[8,9,5,7],[5,8,9,1]],[[5,8,9,4],[8,9,1,4],[8,9,1,4]]],dtype=float)
print(a)   #it will give as float type values
print(a.dtype)
```

```python
#arange it is work as range from number to number
a=np.arange(1,20,3)
print(a)
b=np.arange(20,8,-2)   #reverse printing
print(b)
```

```python
#reshape it is use for convert the number of elements in array list into rows and columns
a=np.arange(1,11,2).reshape(1,5)
print(a)
b=np.array([4,8,9,7,6,3,2,1,5,6,6,8,8,9,6,8])
c=b.reshape(2,4,2)
print(c)
```

```python
#ones,zeros,full
#ones
a=np.ones((2,2,4)).reshape(4,4)
print(a)
b=np.zeros((2,2))
print(b)
c=np.full((2,2),5)   #it use to print any perticular elements except zero  and one
print(c)
```
